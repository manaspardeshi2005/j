```python

target_sequence = """
MQIFVKTLTGKTITLEVESSDTIDNVKSKIQDKEGIPPDQQRLIFAGKQLEDGRTLSDYNIQKESTLHLVLRLRGG
"""
#this is yeast ubiquitin and this is the target we want to modelize it using a template which matches it the most and that template will be human's
#our template will be human ubiquitin
```

```python

# Clean the sequence
target_sequence = "".join(target_sequence.split()).upper()
```

```python

# Validate sequence
valid = set("ACDEFGHIKLMNPQRSTVWY")
if not target_sequence or not set(target_sequence).issubset(valid):
    raise ValueError("Invalid protein sequence!")
```

```python
# Basic protein analysis
from Bio.SeqUtils.ProtParam import ProteinAnalysis
protein = ProteinAnalysis(target_sequence)

print("TARGET PROTEIN ANALYSIS")
print("-----------------------")
print("Sequence:", target_sequence)
print("Length:", len(target_sequence), "amino acids")
print("Molecular Weight:", round(protein.molecular_weight(), 2), "Da")
print("Isoelectric Point:", round(protein.isoelectric_point(), 2))
```

```python
# Save sequence in FASTA format
with open("target.fasta", "w") as file:
    file.write(">Target_Yeast_Ubiquitin\n")
    file.write(target_sequence + "\n")

print("\nFASTA file created successfully!")
```

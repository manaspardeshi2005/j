```python
# Input DNA sequence
#ATGCGTACGTTAGCGATCGATGCGCGATATGCTAGCTAGCTAA
dna = input("Enter DNA sequence: ").upper().replace(" ", "")
```

```python
# Input motif
motif = input("Enter motif to search: ").upper()
```

```python
# 1. Find motif
#ATG
positions = []
start = 0

while True:
    pos = dna.find(motif, start)

    if pos == -1:
        break

    positions.append(pos + 1)
    start = pos + 1

print("\nMotif positions:", positions)
```

```python
# 2. Calculate GC content
gc = dna.count("G") + dna.count("C")
gc_content = (gc / len(dna)) * 100

print("GC Content:", round(gc_content, 2), "%")
```

```python
# 3. Identify coding regions (ORFs)
stop_codons = ["TAA", "TAG", "TGA"]
found = False

for frame in range(3):
    for i in range(frame, len(dna) - 2, 3):

        if dna[i:i+3] == "ATG":

            for j in range(i + 3, len(dna) - 2, 3):

                if dna[j:j+3] in stop_codons:
                    print("\nCoding Region:", dna[i:j+3])
                    print("Start:", i + 1, "End:", j + 3)
                    found = True
                    break

if not found:
    print("\nNo complete coding region found.")
```

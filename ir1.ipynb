```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
text="""
Information retrieval is the process of obtaining relevant information from a collection of information resources.
Search engines use information retrieval techniques to find relevant documents for users.
"""
with open ("document.txt","w") as file:
    file.write(text)
print("Document created successfully")
```

```python
#convert to lower case
text=text.lower()
print("AFter lower case conversion:")
print(text)
```

```python
#tokenization
tokens=word_tokenize(text)
print("tokens:")
print(tokens)
```

```python
#remove punctation
tokens_wo_punc=[]
for token in tokens:
    if token not in string.punctuation:
        tokens_wo_punc.append(token)
print(tokens_wo_punc)
```

```python
# load stop words
stop_words=set(stopwords.words('english'))
print("no of stopwords:",len(stop_words))
print(list(stop_words)[:20])
```

```python
# stopword removal
filtered_tokens=[]
for token in tokens_wo_punc:
    if token not in stop_words:
        filtered_tokens.append(token)
print(filtered_tokens)
```

```python
#apply stemming
stemmer=PorterStemmer()
stemmed_words=[]
for word in filtered_tokens:
    stemmed_words.append(stemmer.stem(word))
print(stemmed_words)
```

```python
#final preprocessed text
final=" ".join(stemmed_words)
print(final)
```

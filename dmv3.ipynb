```python
import pandas as pd
```

```python
df = pd.read_csv("RealEstate_Prices.csv")
```

```python
# 2. Standardize text and remove duplicates
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip().str.lower()

df = df.drop_duplicates()
```

```python
# 3. Handle missing values: mode for text, median for numbers
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())
```

```python
# 4. Filter properties by the most common property type
subset = df[df["property_type"] == df["property_type"].mode()[0]]
print("\nFiltered Data:\n", subset.head())
```

```python
# 5. Aggregate average sale price
print("\nAverage Price by Neighborhood:")
print(df.groupby("neighborhood")["sale_price"].mean())
print("\nAverage Price by Property Type:")
print(df.groupby("property_type")["sale_price"].mean())
```

```python

# 6. Handle price outliers using IQR
q1 = df["sale_price"].quantile(0.25)
q3 = df["sale_price"].quantile(0.75)
iqr = q3 - q1

df["sale_price"] = df["sale_price"].clip(
    q1 - 1.5 * iqr, q3 + 1.5 * iqr
)
```

```python

# 7. Encode categorical columns

df = df.drop(columns=["transaction_date"])

df = pd.get_dummies(df, columns=["property_type"])
```

```python

# 8. Save cleaned dataset
df.to_csv("RealEstate_Cleaned.csv", index=False)
print("\nCleaned dataset saved!")
```

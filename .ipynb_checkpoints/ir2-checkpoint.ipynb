```python
#input documents
n=int(input("enter number of docs:"))
documents={}

for i in range(n):
    documents[f"Document{i+1}"]=input(f"Enter Document{i+1}:").lower().split()

#build inv index
inverted_index={}

for doc,words in documents.items():
    for word in set(words):
        if word not in inverted_index:
            inverted_index[word]=[]
        inverted_index[word].append(doc)

#search query
while True:
    term=input("enter term to search(exit to stop):").lower()
    if term=="exit":
        break
    print(term,"->",",".join(inverted_index.get(term,["NO doc found"])))
```
